```python
archivo = open("nombres.txt" , mode="r") 
lineas = archivo.readlines()

for i, linea in enumerate(lineas, start=1):
    print(i , ".", linea)

archivo.close()
```

```python
archivo = open("frutas.txt", mode="r")

for linea in archivo:
    print("Me gustan las:", linea.strip())

archivo.close()
```

```python
archivo = open("articulo.txt", mode="r")

for linea in archivo:
    if "Python" in linea:
        print(linea)

archivo.close()
```

```python
archivo = open("numeros.txt", mode="r")

suma_total = 0

for linea in archivo:
    suma_total += int(linea)

archivo.close()

print("La suma de los numeros es:", suma_total)
```
