Extra 1: Control de aforo con break y continue (while)
Enunciado: Crea un programa que procese visitantes del museo con un ciclo
`while`. Por cada visitante se captura el costo de su boleto ($0 para menores de 3
años, $30 para 3 a 17 años, $45 para mayores de 18). Si el boleto es gratis,
muestra "Menor de 3 años, sin cargo" y continúa con el siguiente visitante (usa
`continue`). Si el total acumulado alcanza o supera $100, detén el ciclo con
`break` y muestra el total.
Entrada:
Costo del boleto 1: 45
Costo del boleto 2: 0
Costo del boleto 3: 30
Costo del boleto 4: 45
Salida:
Menor de 3 años, sin cargo
Aforo alcanzado, se detiene el registro
Total acumulado: $120.00

Explicación: 
Se ingresan los datos de los visitantes y el programa determina el costo de las entradas de acuerdo con su edad. Al final obtiene el total a pagar.

```python
#Extra 1 
#Control de aforo con break y continue 

total_acumulado = 0
while True:
    costo = float(input("Ingrese el costo del boleto: "))
    #si el boleto es gratuito 
    if costo == 0:
        print("Menos de 3 años, sin cargo")
        #pasamos con el siguiente visitante 
        continue
    #Acumulamos los costos 
    total_acumulado += costo
    #Verificamos si alcanzo el limite 
    if total_acumulado >= 100:
        print("Aforo alcanzado, se detiene el registro")
        break
    print("Total acumulado: $(total_acumulado: .2f )")
```

Extra 2: Estadística de visitantes con for
Enunciado: Crea un programa que pida el número total de visitantes y, con un
ciclo `for`, capture la edad de cada uno. Debe contar cuántos son adultos (18 años
o más) y calcular el promedio de edad de todos los visitantes, mostrando ambos
resultados con dos decimales.
Entrada:
Número de visitantes: 4
Edad del visitante 1: 30
Edad del visitante 2: 45
Edad del visitante 3: 15
Edad del visitante 4: 2
Salida:
Adultos: 2
Promedio de edad: 23.00


Explicación:
El usuario introduce los valores que desea utilizar y selecciona una operación. El programa realiza el cálculo y presenta el resultado.

```python
num_visitantes = int(input("Número de visitantes: "))
cont_adultos = 0
suma_edades = 0

for i in range(1, num_visitantes + 1):
    edad = int(input(f"Edad del visitante {i}: "))
    suma_edades += edad
    if edad >= 18:
        cont_adultos += 1

promedio = suma_edades / num_visitantes

print(f"Adultos: {cont_adultos}")
print(f"Promedio de edad: {promedio:.2f}")



```

Extra 4: Pirámide de asteriscos con for
Enunciado: Crea un programa que pida la altura de una pirámide y la dibuje con
asteriscos usando ciclos `for` anidados y el parámetro `end` de `print`.
Entrada:
Altura de la pirámide: 4
Salida:


Explicación:
El programa pide la altura de la pirámide y usa dos ciclos for: uno para las filas y otro para imprimir los asteriscos. Con end="" los asteriscos se quedan en la misma línea. Así se va formando la pirámide hasta llegar a la altura indicada.

```python
altura = int(input("Altura de la pirámide: "))
for fila in range(1, altura + 1):
   
    for asterisco in range(fila):
        print("*", end="")  
    print()  
```
