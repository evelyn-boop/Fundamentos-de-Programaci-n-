Extra 1: Suma de todos los elementos de una matriz
Enunciado: Crea una función suma_matriz(matriz) que regrese valor (con return) la suma de todos los elementos de una lista de listas, usando ciclos for anidados. Define además una función imprimir_matriz(matriz) que no regrese valor y muestre la matriz sin corchetes ni comas. Aplica ambas funciones a la matriz [[1, 2, 3], [4, 5, 6], [7, 8, 9]].

Explicación de la lógica aplicada:
Para este ejercicio utilicé dos funciones principales. Primero definí imprimir_matriz(), la cual no usa return; en ella recorro con un primer ciclo for cada fila y con un segundo ciclo cada número dentro de la fila, imprimiéndolos con un tabulador \t para que no se muestren comas ni corchetes. Luego creé la función suma_matriz(), donde inicializo un acumulador suma = 0 y, mediante dos ciclos for anidados, voy recorriendo cada celda de la matriz para sumar sus datos y al final devolver el total acumulado usando return.

```python
# Función para imprimir la matriz formateada (sin return)
def imprimir_matriz(matriz):
    for fila in matriz:
        for elemento in fila:
            print(elemento, end="\t")
        print()

# Función para calcular la suma total (con return)
def suma_matriz(matriz):
    suma = 0
    for fila in matriz:
        for elemento in fila:
            suma += elemento
    return suma

# Código principal
matriz = [[1,2,3], 
          [4,5,6], 
          [7,8,9]]

imprimir_matriz(matriz)
resultado = suma_matriz(matriz)
print("La suma de todos los elementos de la matriz es:", resultado)
```

Extra 2: Matriz con suma por fila
Enunciado: Crea una función suma_fila(fila) que regrese valor (con return) la suma de los elementos de una lista, y una función mostrar_tabla(matriz) que no regrese valor e imprima la matriz sin corchetes ni comas. Dada la matriz [[3, 1, 4], [1, 5, 9], [2, 6, 5]], imprime la matriz y la suma de cada fila.

Explicación de la lógica aplicada:
En este problema dividí el proceso en dos partes: la función mostrar_tabla() se encarga de imprimir la matriz línea por línea usando \t para quitar los corchetes de la visualización. La función suma_fila() recibe únicamente una lista (una fila) y suma sus elementos para regresar el resultado individual con return. En el programa principal, recorro los índices de las filas con un ciclo range(len(matriz)) y mando a llamar a suma_fila() enviando una fila a la vez para mostrar el resultado correspondiente de cada posición

```python
# Función para sumar una sola fila (con return)
def suma_fila(fila):
    suma = 0
    for numero in fila:
        suma += numero
    return suma

# Función para imprimir la matriz sin corchetes (sin return)
def mostrar_tabla(matriz):
    for fila in matriz:
        for elemento in fila:
            print(elemento, end="\t")
        print()

# Código principal
matriz = [[3, 1, 4],
          [1, 5, 9], 
          [2, 6, 5]]

mostrar_tabla(matriz)

for i in range(len(matriz)):
    total_fila = suma_fila(matriz[i])
    print(f"Suma de la fila {i}: {total_fila}")
```

Extra 3: Producto de una columna sin usar el operador *
Enunciado:Crea la matriz [[1, 2, 3], [4, 5, 6], [7, 8, 9], [2, 4, 6]] (4 filas y 3 columnas). Define una función multiplicar_columna(matriz, columna) que regrese valor (con return) el producto de los elementos de una columna dada. Restricción: dentro de la función está prohibido usar el operador *; el producto debe calcularse con sumas repetidas usando un ciclo. Pide al usuario el número de columna (0, 1 o 2) y muestra el resultado.

Explicación de la lógica aplicada:
Como no se podía usar el operador de multiplicación *, apliqué la lógica de multiplicaciones mediante sumas acumuladas. La función multiplicar_columna() empieza con producto = 1. Al recorrer los elementos de la columna elegida por el usuario, uso un ciclo interno que suma la variable producto tantas veces como lo indique el número en esa posición. Al terminar el recorrido de todas las filas, la función regresa el resultado con return. También agregué una validación simple con if para asegurarme de que el usuario ingrese un índice de columna válido (0, 1 o 2).

```python
# Función para calcular el producto de una columna con sumas (con return)
def multiplicar_columna(matriz, columna):
    producto = 1
    for fila in matriz:
        numero = fila[columna]
        acumulado = 0
        # Sumas repetidas para sustituir el uso de *
        for i in range(numero):
            acumulado += producto
        producto = acumulado
    return producto

# Código principal
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
    [2, 4, 6]
]

col = int(input("Numero de columna: "))

if 0 <= col <= 2:
    resultado = multiplicar_columna(matriz, col)
    print(f"El producto de la columna {col} es: {resultado}")
else:
    print("Número de columna fuera de rango (debe ser 0, 1 o 2).")
```

Extra 4: Tabla de Pitágoras de tamaño variable
Enunciado:Crea una función generar_tabla(tamano) que regrese valor (con return) una lista de listas de tamano x tamano con los productos de la tabla de Pitágoras (renglones y columnas del 1 al tamano). Dentro de la función de generación está prohibido usar el operador *; usa sumas repetidas con ciclos. Define una función imprimir_tabla(tabla) que no regrese valor e imprima la tabla sin corchetes ni comas, con los números tabulados (\t). Pide al usuario un tamaño entre 2 y 5; si ingresa un valor fuera del rango, muestra un mensaje de error. Al final pide un renglón y una columna y muestra el producto consultando la matriz, sin usar *.

Explicación de la lógica aplicada:
Para resolver este problema construí la matriz dinámicamente con la función generar_tabla(). Dentro de ella creé las filas con append() calculando la multiplicación de renglón por columna mediante un ciclo de sumas repetidas para respetar la regla de no usar *. Luego, la función imprimir_tabla() se encarga de mostrar la matriz limpia con tabuladores. En el programa principal, valido que el tamaño esté entre 2 y 5 y que los factores pedidos no se salgan de los límites. Finalmente, en lugar de hacer operaciones matemáticas para la respuesta, obtengo el valor directamente consultando la posición de la matriz matriz[renglon - 1][columna - 1].

```python
# Función para construir la tabla sin usar * (con return)
def generar_tabla(tamano):
    tabla = []
    for r in range(1, tamano + 1):
        fila = []
        for c in range(1, tamano + 1):
            producto = 0
            for i in range(c):
                producto += r
            fila.append(producto)
        tabla.append(fila)
    return tabla

# Función para desplegar la tabla (sin return)
def imprimir_tabla(tabla):
    for fila in tabla:
        for valor in fila:
            print(valor, end="\t")
        print()

# Código principal
tamano = int(input("Tamano de la tabla: "))

if 2 <= tamano <= 5:
    renglon = int(input("Renglon (factor): "))
    columna = int(input("Columna (factor): "))
    
    if 1 <= renglon <= tamano and 1 <= columna <= tamano:
        matriz = generar_tabla(tamano)
        imprimir_tabla(matriz)
        
        # Consulta de resultado mediante posicionamiento en la matriz
        resultado = matriz[renglon - 1][columna - 1]
        print(f"El producto de {renglon} x {columna} es: {resultado}")
    else:
        print("Los factores indicados están fuera del rango de la tabla.")
else:
    print("El tamaño ingresado debe estar entre 2 y 5.")
```
