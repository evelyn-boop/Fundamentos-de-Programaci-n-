Extra 1: Suma de todos los elementos de una matriz
Enunciado: Crea una función suma_matriz(matriz) que regrese valor (con return) la suma de todos los elementos de una lista de listas, usando ciclos for anidados. Define además una función imprimir_matriz(matriz) que no regrese valor y muestre la matriz sin corchetes ni comas. Aplica ambas funciones a la matriz [[1, 2, 3], [4, 5, 6], [7, 8, 9]].

Explicación de la lógica aplicada:
Para este ejercicio utilicé dos funciones principales. Primero definí imprimir_matriz(), la cual no usa return; en ella recorro con un primer ciclo for cada fila y con un segundo ciclo cada número dentro de la fila, imprimiéndolos con un tabulador \t para que no se muestren comas ni corchetes. Luego creé la función suma_matriz(), donde inicializo un acumulador suma = 0 y, mediante dos ciclos for anidados, voy recorriendo cada celda de la matriz para sumar sus datos y al final devolver el total acumulado usando return.

```python
# Función para imprimir la matriz formateada (sin return)
def imprimir_matriz(matriz):
    for fila in matriz:
        for elemento in fila:
            print(elemento, end="\t")
        print()

# Función para calcular la suma total (con return)
def suma_matriz(matriz):
    suma = 0
    for fila in matriz:
        for elemento in fila:
            suma += elemento
    return suma

# Código principal
matriz = [[1,2,3], 
          [4,5,6], 
          [7,8,9]]

imprimir_matriz(matriz)
resultado = suma_matriz(matriz)
print("La suma de todos los elementos de la matriz es:", resultado)
```
