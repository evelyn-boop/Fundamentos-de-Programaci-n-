Algoritmo para calcular velocidad 

```python
distancia= int(input("Ingrese la distancia en kilometros:"))
tiempo= int(input("Ingrese el tiempo en horas: "))

print("el dato distancia es de tipo: ",type(distancia))
print("el dato tiempo es de tiempo: ",type(tiempo))
velocidad = distancia/tiempo
print("La velocidad promedio es:", velocidad, "km/h")
```

conversión de temperaturas 

```python

celsius = float(input("Ingresa la temperatura en grados Celsius: "))
fahrenheit = (celsius * 9/5) + 32
kelvin = celsius + 273.15

print(f"\nTemperatura en Fahrenheit: {fahrenheit:.2f} °F")
print(f"Temperatura en Kelvin: {kelvin:.2f} K")
```

calculo promedio de evaluaciones 

```python
evaluacion1 = float(input("Ingresa la calificación de la primera evaluación: "))
evaluacion2 = float(input("Ingresa la calificación de la segunda evaluación: "))
evaluacion3 = float(input("Ingresa la calificación de la tercera evaluación: "))

promedio = (evaluacion1 + evaluacion2 + evaluacion3) / 3

print(f"\nEl promedio de las evaluaciones es: {promedio:.2f}")
```
