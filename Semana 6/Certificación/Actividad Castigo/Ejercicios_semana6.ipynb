Ejercicio 1: Área de un triángulo
Escribe un programa en Python que aplique el modelo Entrada-Proceso-Salida para calcular el área de un triángulo utilizando la fórmula:
area = (base * altura) / 2
El programa debe utilizar input(), float() y print(). La entrada debe ser una base de 8 y una altura de 5.
Explicación:
En este ejercicio primero pedimos la base y la altura del triángulo. Después hacemos la operación para obtener el área y finalmente mostramos el resultado en pantalla.

```python
# ==========================================
# EJERCICIO 1: Área de un triángulo
# ==========================================

base = float(input("Ingresa la base del triángulo: "))
altura = float(input("Ingresa la altura del triángulo: "))

area = (base * altura) / 2

print("El área del triángulo es:", area)
```

Ejercicio 10: Suma acumulada con while
Pide números mientras sean diferentes de 0. El número 0 debe terminar el ciclo y no debe sumarse. La entrada es 5, 8, 2 y 0.
Explicación:
Usamos un ciclo while para seguir pidiendo números. Cada número se va agregando a una suma y cuando se introduce 0, el ciclo termina.

```python
# ==========================================
# EJERCICIO 10: Suma acumulada
# ==========================================

suma = 0

numero = int(input("Ingresa un número (0 para terminar): "))

while numero != 0:
    suma = suma + numero
    numero = int(input("Ingresa otro número (0 para terminar): "))

print("La suma de los números es:", suma)
```

Ejercicio 12: Suma de pares del 1 al 20 con for y continue
Crea un programa que recorra los números del 1 al 20, identifique los números pares utilizando un ciclo for y continue, y calcule la suma de todos los números pares.
Explicación:
Recorremos los números del 1 al 20. Cuando encontramos un número impar usamos continue para saltarlo y solamente sumamos los números pares.

```python
# ==========================================
# EJERCICIO 12: Suma de números pares
# ==========================================

suma = 0

for numero in range(1, 21):

    if numero % 2 != 0:
        continue

    suma = suma + numero

print("La suma de los números pares del 1 al 20 es:", suma)
```

Ejercicio 13: Suma de una lista con función
Crea una función que reciba una lista y regrese la suma de sus elementos sin utilizar la función sum(). La lista es:
[10, 20, 30, 40]
Explicación:

```python
# ==========================================
# EJERCICIO 13: Suma de una lista
# ==========================================

def sumar_lista(lista):
    suma = 0

    for numero in lista:
        suma = suma + numero

    return suma


numeros = [10, 20, 30, 40]

resultado = sumar_lista(numeros)

print("La suma de los elementos es:", resultado)
```

Ejercicio 14: Máximo de una lista
Crea una función que reciba una lista y regrese el elemento máximo sin utilizar la función max(). La lista es:
[7, 3, 9, 2]
Explicación:
Comenzamos tomando el primer elemento como el mayor. Después recorremos la lista y comparamos los demás elementos para encontrar cuál tiene el valor más grande.

```python
# ==========================================
# EJERCICIO 14: Máximo de una lista
# ==========================================

def encontrar_maximo(lista):
    mayor = lista[0]

    for numero in lista:
        if numero > mayor:
            mayor = numero

    return mayor


numeros = [7, 3, 9, 2]

resultado = encontrar_maximo(numeros)

print("El elemento máximo es:", resultado)
```

Ejercicio 15: Matriz y suma de la diagonal principal
Crea una matriz de 3 × 3 con los valores:
[[1,2,3],[4,5,6],[7,8,9]]
Después calcula la suma de los elementos que forman la diagonal principal.
Explicación:
La diagonal principal está formada por los elementos que tienen la misma posición de fila y columna. En este caso son 1, 5 y 9, por lo que la suma es 15.

```python
# ==========================================
# EJERCICIO 15: Diagonal principal
# ==========================================

matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

suma = 0

for i in range(len(matriz)):
    suma = suma + matriz[i][i]

print("La suma de la diagonal principal es:", suma)
```

Ejercicio 16: Promedio con función que no regresa valor
Crea una función que reciba una lista de calificaciones y calcule el promedio. La función debe mostrar directamente el resultado y no regresar un valor. La lista es:
[8, 9, 10, 7]
Explicación:
La función suma las calificaciones y las divide entre la cantidad de elementos. Después imprime el promedio utilizando dos decimales.

```python
# ==========================================
# EJERCICIO 16: Promedio
# ==========================================

def calcular_promedio(calificaciones):
    suma = 0

    for calificacion in calificaciones:
        suma = suma + calificacion

    promedio = suma / len(calificaciones)

    print(f"El promedio es: {promedio:.2f}")


calificaciones = [8, 9, 10, 7]

calcular_promedio(calificaciones)
```

Ejercicio 18: Factorial con función recursiva
Crea una función recursiva que calcule el factorial de un número. La entrada es 5 y el resultado esperado es 120.
Explicación:
Una función recursiva es una función que se llama a sí misma. En este caso va multiplicando el número por el factorial del número anterior hasta llegar a 1.

```python
# ==========================================
# EJERCICIO 18: Factorial recursivo
# ==========================================

def factorial(numero):

    if numero == 0 or numero == 1:
        return 1

    return numero * factorial(numero - 1)


numero = int(input("Ingresa un número: "))

resultado = factorial(numero)

print("El factorial de", numero, "es:", resultado)
```

Ejercicio 19: Tupla: acceso y suma con función
Utiliza la siguiente tupla:
(4, 8, 15, 16, 23, 42)
El programa debe mostrar el tercer elemento y calcular la suma de todos los elementos mediante una función.
Explicación:
En este ejercicio practicamos el uso de tuplas. Para obtener el tercer elemento utilizamos el índice 2, ya que Python comienza a contar desde cero. También hacemos una función para sumar sus elementos.

```python
# ==========================================
# EJERCICIO 19: Tupla
# ==========================================

def sumar_tupla(tupla):
    suma = 0

    for numero in tupla:
        suma = suma + numero

    return suma


numeros = (4, 8, 15, 16, 23, 42)

print("El tercer elemento es:", numeros[2])

resultado = sumar_tupla(numeros)

print("La suma de la tupla es:", resultado)
```

Ejercicio 21: Diccionario de calificaciones y promedios
Crea un diccionario que contenga las calificaciones de Ana y Beto:
{"Ana": [8, 9, 10], "Beto": [7, 7, 8]}
El programa debe calcular y mostrar el promedio de cada estudiante.
Explicación:
Cada nombre tiene una lista con sus calificaciones. Recorremos el diccionario y calculamos el promedio de cada lista para mostrarlo con dos decimales.

```python
# ==========================================
# EJERCICIO 21: Calificaciones y promedios
# ==========================================

calificaciones = {
    "Ana": [8, 9, 10],
    "Beto": [7, 7, 8]
}

for nombre, notas in calificaciones.items():

    suma = 0

    for nota in notas:
        suma = suma + nota

    promedio = suma / len(notas)

    print(f"{nombre}: {promedio:.2f}")
```
