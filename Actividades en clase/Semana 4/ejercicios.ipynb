```python

n = int(input("Ingresa el tamaño de la matriz cuadrada: "))

matriz = []
for f in range(n):
    fila = [0] * n
    matriz.append(fila)
for f in range(n):
    for c in range(n):
        if f + c == n - 1:
            matriz[f][c] = f + 1  

print("\n Matriz con diagonal invertida:")
for f in range(n):
    for c in range(n):
        print(matriz[f][c], end="  ")
    print()
```

```python
dimension = int(input("Cuántas dimensiones: "))

matriz = []
for f in range(dimension):
    fila_actual = []
    for c in range(dimension):
        if f == c or f + c == dimension -1:
            valor = 1
        else:
            valor = 0
        fila_actual.append(valor)
    matriz.append(fila_actual)

print("Impresión de matriz")
for fila in matriz:
    for elemento in fila:
        print(elemento, end="  ")
    print()
```

```python
# Listas vacías
positivos = []
negativos = []

# Pedir 10 números positivos
print("Ingresa 10 números positivos:")
while True:
    if valor(positivos) >= 10: 
        break
    numero = int(input("Número: "))
    if numero > 0:
        positivos.append(numero)
    else:
        print("Error, debe ser positivo. Intenta otra vez.")

# Pedir 10 números negativos
print("\nIngresa 10 números negativos:")
while True:
    if valor(negativos) >= 10:
        break
    numero = int(input("Número: "))
    if numero < 0:
        negativos.append(numero)
    else:
        print("Error, debe ser negativo. Intenta otra vez.")

# Mostrar resultados
print("\n POSITIVOS  |  NEGATIVOS ")
for i in range(10):
    print(f"   {positivos[i]:>2}        {negativos[i]:>3}")
```

```python
lista =[]
contador = 1
while contador <= 10:
    valor=int(input("Introduce un número entre -10 y 10:"))
    if valor >= -10 and valor <=10:
        contador +=1
        lista.append(valor)
    else:
          print("Rango incorrecto, ingresa un valor entre -10 y 10")

print(lista)
for i in lista:
     if i <0:
          print("asteriscos negativos")
     elif i > 0:
          print("asteriscos positivos")
     else:
          print("|")

for i in lista:
    if i < 0:
        print(("*" * abs(i)) + "       |    ")
    elif i > 0:
        print("         |   " + ("*" * i))
    
```
