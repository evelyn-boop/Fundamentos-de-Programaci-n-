```python
filas = 3
columnas = 3

# dato = int(input("Ingresa el valor de tu matriz: "))

matriz = []
for f in range(filas):
    fila_actual = []
    for c in range(columnas):
        # matriz [f] [c] = valor = int(input(f"Valor para [{f}] [{c}] "))
        valor = int(input(f"Valor para [{f}] [{c}] "))
        fila_actual.append(valor)
        print(fila_actual)
    matriz.append(fila_actual)

print("Impresión de matriz")
for f in range(filas):
    for c in range(columnas):
        print(matriz[f][c], end=" ")
    print()
```

```python

n = int(input("Ingresa el tamaño de la matriz cuadrada: "))

matriz = []
for f in range(n):
    fila = [0] * n
    matriz.append(fila)
for f in range(n):
    for c in range(n):
        if f + c == n - 1:
            matriz[f][c] = f + 1  

print("\n Matriz con diagonal invertida:")
for f in range(n):
    for c in range(n):
        print(matriz[f][c], end="  ")
    print()
```

```python
dimension = int(input("Cuántas dimensiones: "))

matriz = []
for f in range(dimension):
    fila_actual = []
    for c in range(dimension):
        if f == c or f + c == dimension -1:
            valor = 1
        else:
            valor = 0
        fila_actual.append(valor)
    matriz.append(fila_actual)

print("Impresión de matriz")
for fila in matriz:
    for elemento in fila:
        print(elemento, end="  ")
    print()
```
