Ejercicio 1 

```python
horas = int(input("¿Cuántas horas estuvo el carro? "))

if horas <= 2:
    total = horas * 15
else:
    total = (2 * 15) + ((horas - 2) * 10)

if total > 200:
    total = 200

print(f"Total a pagar: ${total:.2f}")

```

Ejercicio 2

```python
inicio = int(input("Escribe el año inicial: "))
fin = int(input("Escribe el año final: "))

print("Años bisiestos:", end=" ")

for año in range(inicio, fin + 1):
    if (año % 4 == 0 and año % 100 != 0) or año % 400 == 0:
        print(año, end=" ")

print()
```

Ejercicio 3

```python
numero = int(input("Escribe un número entero positivo: "))

original = numero
contador = 0

while numero > 0:
    numero = numero // 10
    contador = contador + 1

print(f"El número {original} tiene {contador} dígitos")
```

Ejercicio 4

```python
numero = int(input("Escribe un número entero positivo: "))

original = numero
suma = 0

while numero > 0:
    digito = numero % 10
    suma = suma + digito
    numero = numero // 10

print(f"La suma de los dígitos de {original} es: {suma}")
```

Ejercicio 5 

```python
numero = int(input("Escribe un número entero positivo: "))

original = numero
invertido = 0

while numero > 0:
    digito = numero % 10
    invertido = invertido * 10 + digito
    numero = numero // 10

print(f"El número invertido de {original} es {invertido}")
```

Ejercicio 6

```python
dividendo = int(input("Escribe el dividendo: "))
divisor = int(input("Escribe el divisor: "))

cociente = 0
residuo = dividendo

while residuo >= divisor:
    residuo = residuo - divisor
    cociente = cociente + 1

print(f"{dividendo} ÷ {divisor} = {cociente} con residuo {residuo}")

```

Ejercicio 9

```python
cantidad = int(input("¿Cuántos números vas a capturar? "))

mayor = 0

for i in range(cantidad):
    numero = int(input(f"Escribe el número {i + 1}: "))

    if i == 0:
        mayor = numero
    elif numero > mayor:
        mayor = numero

print("El número mayor es:", mayor)
```

Ejercicio 11

```python
n = int(input("Escribe un número positivo: "))

resultado = 0

for numero in range(1, n + 1):

    if numero % 2 == 0:
        resultado = resultado - numero
    else:
        resultado = resultado + numero

print("El resultado de la serie es:", resultado)
```

Ejercicio 12

```python
n = int(input("Escribe un número positivo: "))

suma = 0.0

for numero in range(1, n + 1):
    suma = suma + (1 / numero)

print(f"La suma de la serie es: {suma:.2f}")
```

Ejercicio 13

```python
n = int(input("¿Cuántos términos quieres mostrar? "))

a = 0
b = 1

print("Fibonacci:", end=" ")

for i in range(n):
    print(a, end=" ")

    siguiente = a + b
    a = b
    b = siguiente

print()
```

Ejercicio 14

```python
filas = int(input("¿Cuántas filas quieres? "))

for i in range(1, filas + 1):

    linea = ""

    for j in range(i):
        linea = linea + "*"

    print(linea)
```
