Una **tupla** es una secuencia ordenada e **inmutable** de elementos en Python. A
diferencia de las listas, una vez creada una tupla no se pueden agregar, eliminar
ni modificar sus elementos. Se define con paréntesis `()`.

```python
def suma_tupla(t):
    total=0
    for e in t:
        total += e
    return total

calorias=(800,950,700,1300,800,900,650)
resultado = suma_tupla(calorias)
print("Metiste a tu cuerpo {resultado}calorias,gordo!")

```

```python
candidatos={
    "A" :0,
    "B":0,
    "C":0
}

cantidad=int(input("¿Cuántos votos van a ser"))
conteo=1
#for c in candidatos.keys():
    #print(c)


while conteo <= cantidad:
    voto=input("Ingresa el voto {conteo} (A,B,C): ")
    if voto.upper() in candidatos:
        print(f"Votos  {conteo}")

        print(f"Voto para {voto.upper()}")
        candidatos[voto.upper()] +=1
        conteo +=1
else:
    print("El voto no es válido, vota nuevamente")

print("\n Resultados:")
for k,v in candidatos.items():
    print(f"Opción {k}, votos: {v}")
```
