Ejercicio 1: ### Extra 1: Sistema de calificaciones con tuplas
**Enunciado:** Crea una tupla `calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)`.
Imprime la tercera calificación, captura dos calificaciones nuevas con `input()` y
anéxalas en una nueva tupla. Convierte la nueva tupla en lista, ordénala **de mayor
a menor** y muestra la suma de todos los elementos usando una función que retorne
el valor.
**Entrada:**
```
Nueva calificación 1: 8.5
Nueva calificación 2: 9.5
```
**Salida:**
```
Tercera calificación: 8.0
Nueva tupla: (7.5, 9.0, 8.0, 6.5, 10.0, 8.5, 9.5)
Lista ordenada (mayor a menor): [10.0, 9.5, 9.0, 8.5, 8.0, 7.5, 6.5]
Suma total: 59.0


Explicación: Acceso por índice: Creamos la tupla calificaciones e imprimimos el tercer elemento usando la posición [2].
Concatenación: Capturamos dos notas con input(), las convertimos a decimales (float) y creamos una nueva tupla sumando la original con (nota1, nota2).
Lista y orden descendente: Convertimos la nueva tupla a lista mediante list() y la ordenamos de mayor a menor con el método .sort(reverse=True).
Función con retorno: Definimos la función sumar_notas() que utiliza sum() para calcular el total de los elementos de la tupla y retorna el resultado para mostrarlo en pantalla.

````python
# Definición de la función
def sumar_notas(tupla):
    return sum(tupla)

# 1. Tupla inicial y tercer elemento
calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)
print("Tercera calificación:", calificaciones[2])

# 2. Captura de datos y nueva tupla
nota1 = float(input("Nueva calificación 1: "))
nota2 = float(input("Nueva calificación 2: "))
nueva_tupla = calificaciones + (nota1, nota2)
print("Nueva tupla:", nueva_tupla)

# 3. Lista ordenada de mayor a menor
lista_notas = list(nueva_tupla)
lista_notas.sort(reverse=True)
print("Lista ordenada (mayor a menor):", lista_notas)

# 4. Función y suma total
suma_total = sumar_notas(nueva_tupla)
print("Suma total:", suma_total)
````

Ejercicio 2:
### Extra 2: Agenda de contactos con búsqueda
**Enunciado:** Crea un diccionario `agenda` con tres contactos iniciales (`"Ana":
"555-0101"`, `"Luis": "555-0102"`, `"Mía": "555-0103"`). Captura un nuevo contacto
(nombre y teléfono) con `input()` y agrégalo al diccionario. Luego imprime todos
los nombres con `keys()` en una sola línea y, con una función que retorne el valor,
busca y muestra el teléfono del nuevo contacto.
**Entrada:**
```
Nombre del nuevo contacto: Pedro
Teléfono del nuevo contacto: 555-0140
Nombre a buscar: Pedro
```
**Salida:**
```
Contactos registrados: Ana, Luis, Mía, Pedro
El teléfono de Pedro es: 555-0140

Explicación: Creación del diccionario: Inicializamos agenda con tres contactos base (clave: nombre, valor: teléfono).
Adición de datos: Solicitamos un nombre y un teléfono con input() para añadir un nuevo elemento mediante agenda[nombre] = telefono.
Lectura de claves: Obtenemos e imprimimos los nombres con el método .keys(), uniéndolos por comas.
Función con retorno: Creamos la función buscar_telefono() que recibe el diccionario y el nombre buscado, retorna el valor del número telefónico y lo despliega en consola.

````python
# Función para buscar teléfono
def buscar_telefono(diccionario, nombre):
    return diccionario.get(nombre)

# 1. Diccionario inicial
agenda = {"Ana": "555-0101", "Luis": "555-0102", "Mía": "555-0103"}

# 2. Agregar nuevo contacto
nuevo_nombre = input("Nombre del nuevo contacto: ")
nuevo_tel = input("Teléfono del nuevo contacto: ")
agenda[nuevo_nombre] = nuevo_tel

# 3. Imprimir nombres registrados
nombres = ", ".join(agenda.keys())
print("Contactos registrados:", nombres)

# 4. Buscar y mostrar teléfono
contacto_buscar = input("Nombre a buscar: ")
telefono_encontrado = buscar_telefono(agenda, contacto_buscar)

if telefono_encontrado:
    print(f"El teléfono de {contacto_buscar} es: {telefono_encontrado}")
else:
    print("Contacto no encontrado.")
````

### Extra 3: Calculadora segura con manejo de excepciones
**Enunciado:** Escribe un programa que pida dos números enteros. Implementa un
bloque `try-except` que capture `ValueError` (caracteres no numéricos o vacíos) con
un mensaje controlado y un bloque específico `except ZeroDivisionError` que muestre
un mensaje amigable si el segundo número es cero. En caso exitoso, muestra la
división del primero entre el segundo.
**Entrada:**
```
Primer número entero: 20
Segundo número entero: 4
```
**Salida:**
```
20 dividido entre 4 es: 5.0
```
**Caso de error (divisor cero):**
```
Primer número entero: 10
Segundo número entero: 0
```
**Salida:**
```
Error: No es posible dividir entre cero. Ingresa un divisor distinto de 0.

Explicación:
Entrada y captura de datos: Pedimos dos enteros al usuario y los convertimos dentro de un bloque try.
Excepción por valor no numérico: Usamos except ValueError para detectar texto, letras o entradas vací.as, arrojando un mensaje de error sin que el programa colapse.
Excepción por división entre cero: Usamos un bloque específico except ZeroDivisionError para capturar cuando el segundo número es 0.
Resultado: Si no hay errores, se calcula y muestra la división en el bloque principal o dentro del flujo normal.

````python
try:
    num1 = int(input("Primer número entero: "))
    num2 = int(input("Segundo número entero: "))
    
    resultado = num1 / num2
    print(f"{num1} dividido entre {num2} es: {resultado}")

except ValueError:
    print("Error: Ingresa únicamente números enteros válidos.")

except ZeroDivisionError:
    print("Error: No es posible dividir entre cero. Ingresa un divisor distinto de 0.")
````

### Extra 4: Analizador de mensajes con strings
**Enunciado:** Crea una variable `mensaje = "Python es un lenguaje poderoso"`.
Imprime su longitud con `len()`, conviértelo a mayúsculas, reemplaza la palabra
`"Python"` por `"programación"` y, con una función que retorne el conteo, muestra
cuántas palabras contiene el mensaje original.
**Entrada:**
```
mensaje = "Python es un lenguaje poderoso"
```
**Salida:**
```
Longitud del mensaje: 30
En mayúsculas: PYTHON ES UN LENGUAJE PODEROSO
Texto reemplazado: programación es un lenguaje poderoso
Palabras totales: 5

Explicación: Longitud y mayúsculas: Usamos len(mensaje) para la longitud y el método .upper() para convertir todo a mayúsculas.
Reemplazo de palabras: Usamos .replace("Python", "programación") para reemplazar la palabra especificada.
Función con retorno: La función contar_palabras() divide el texto con .split() y usa len() sobre la lista resultante para retornar el conteo exacto de palabras.

````python
def contar_palabras(texto):
    palabras = texto.split()
    return len(palabras)

mensaje = "Python es un lenguaje poderoso"

# 1. Longitud
print("Longitud del mensaje:", len(mensaje))

# 2. Mayúsculas
print("En mayúsculas:", mensaje.upper())

# 3. Reemplazo
texto_modificado = mensaje.replace("Python", "programación")
print("Texto reemplazado:", texto_modificado)

# 4. Conteo de palabras mediante función
total_palabras = contar_palabras(mensaje)
print("Palabras totales:", total_palabras)
````

### Desafío Extra 2: Contador de votos con diccionario
Crea un diccionario `votos = {"Rojo": 0, "Azul": 0, "Verde": 0}`. Pide al usuario
el color de su voto cinco veces, suma uno al valor correspondiente en cada ocasión
y, al final, imprime con `items()` el color ganador junto con su total.

Explicación:Inicialización: Declaramos un diccionario votos con contadores en cero para cada color.
Ciclo de votación: Implementamos un ciclo for de 5 repeticiones para pedir la opción del usuario e incrementar en 1 el contador del color correspondiente.
Determinación del ganador: Iteramos sobre votos.items() para encontrar el color que acumuló la cantidad máxima de votos.

````python
votos = {"Rojo": 0, "Azul": 0, "Verde": 0}

print("Opciones de color: Rojo, Azul, Verde")

# Votación 5 veces
for i in range(5):
    eleccion = input(f"Voto {i+1} - Elige un color: ").capitalize()
    if eleccion in votos:
        votos[eleccion] += 1
    else:
        print("Color no válido, el voto no se contó.")

# Determinar ganador iterando con items()
ganador = ""
max_votos = -1

for color, cantidad in votos.items():
    if cantidad > max_votos:
        max_votos = cantidad
        ganador = color

print(f"\nEl ganador es el color {ganador} con {max_votos} votos.")
````

### Desafío Extra 3: Validación robusta de entrada
Escribe un programa que pida al usuario dos números enteros dentro de un bloque
`try-except`. Si el usuario ingresa texto o un valor vacío, muestra un mensaje
controlado y repite la solicitud con un ciclo `while`. Luego pide la división del
primero entre el segundo, con un bloque específico para `ZeroDivisionError`.

Explicación: Validación iterativa: Usamos un ciclo while True junto con un bloque try-except ValueError para pedir los números continuamente hasta que el usuario ingrese datos enteros válidos.
División segura: Una vez capturados correctamente los dos enteros, intentamos realizar la división atrapando ZeroDivisionError si el divisor ingresado fue 0.

````python
# Repetición hasta obtener dos enteros válidos
while True:
    try:
        num1 = int(input("Ingresa el primer entero: "))
        num2 = int(input("Ingresa el segundo entero: "))
        break 
    except ValueError:
        print("Entrada no válida. Por favor ingresa números enteros.\n")

# Operación de división con manejo de divisor cero
try:
    resultado = num1 / num2
    print(f"Resultado de la división: {resultado}")
except ZeroDivisionError:
    print("Error: No es posible dividir entre cero.")
````

Ejercicio 5:
### Desafío Extra 5: Minimenú modular integrador
Construye un menú con un ciclo `while` y opciones numéricas que permita ejecutar de
forma modular: (1) mostrar la tupla de números ordenada, (2) buscar un teléfono en
el diccionario `contactos`, (3) dividir dos números con manejo de
`ZeroDivisionError` y (4) analizar un mensaje de texto. La opción (5) debe
finalizar el programa.

Explicación: Estructura iterativa: Usamos un ciclo while para mantener el menú desplegado de forma interactiva.
Delegación por opciones: Dependiendo de la opción ingresada (1 a 4), se ejecuta una tarea de tuplas, diccionarios, excepciones o strings.
Criterio de salida: La opción 5 rompe el ciclo y finaliza la ejecución de la aplicación de consola.

````python
def mostrar_tupla():
    t = (15, 30, 5, 20)
    lista = list(t)
    lista.sort()
    print("Tupla ordenada:", lista)

def buscar_contacto():
    contactos = {"Juan": "555-1111", "Sofía": "555-2222"}
    nombre = input("Ingresa el nombre a buscar: ")
    print("Teléfono:", contactos.get(nombre, "No encontrado"))

def dividir_numeros():
    try:
        n1 = float(input("Número 1: "))
        n2 = float(input("Número 2: "))
        print("Resultado:", n1 / n2)
    except ZeroDivisionError:
        print("Error: No puedes dividir entre cero.")
    except ValueError:
        print("Error: Ingresa números válidos.")

def analizar_texto():
    m = input("Ingresa un mensaje: ")
    print(f"Palabras totales: {len(m.split())}")

# Menú principal con ciclo while
opcion = ""
while opcion != "5":
    print("\n--- MINIMENÚ INTEGRADOR ---")
    print("1. Ver tupla ordenada")
    print("2. Buscar teléfono")
    print("3. Dividir números")
    print("4. Contar palabras de texto")
    print("5. Salir")
    
    opcion = input("Selecciona una opción (1-5): ")
    
    if opcion == "1":
        mostrar_tupla()
    elif opcion == "2":
        buscar_contacto()
    elif opcion == "3":
        dividir_numeros()
    elif opcion == "4":
        analizar_texto()
    elif opcion == "5":
        print("Saliendo del programa...")
    else:
        print("Opción inválida. Intenta nuevamente.")
````
