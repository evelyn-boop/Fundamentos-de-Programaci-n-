En esta actividad realizamos un programa en Python utilizando tuplas,
diccionarios, excepciones y strings. También utilizamos funciones y un
menú principal para poder seleccionar cada sección del programa.

```python
# ==========================================
# ACTIVIDAD SEMANA 5 - MENU MODULAR
# Nombre: Evelyn Itzel León Uribe
# Matrícula: AL07311179
# ==========================================


# ==========================================
# SECCIÓN 1: TRABAJO CON TUPLAS
# ==========================================

# Función que recibe la tupla y retorna la suma de sus elementos
def sumar_calificaciones(tupla_notas):
    suma_total = sum(tupla_notas)
    return suma_total


def trabajar_tuplas():

    # Crear una tupla con 5 calificaciones
    calificaciones = (85, 90, 78, 92, 88)

    print("\n--- REGISTRO DE CALIFICACIONES ---")
    print("Calificaciones iniciales:", calificaciones)

    # Mostrar el tercer elemento de la tupla
    print("La calificación del tercer parcial es:", calificaciones[2])

    # Capturar dos calificaciones adicionales
    nota1 = float(input("Ingresa la calificación de la Tarea 1: "))
    nota2 = float(input("Ingresa la calificación de la Tarea 2: "))

    # Crear una nueva tupla
    todas_las_notas = calificaciones + (nota1, nota2)

    print("Nueva tupla con todas las notas:", todas_las_notas)

    # Convertir la tupla a lista
    lista_notas = list(todas_las_notas)

    # Ordenar la lista
    lista_notas.sort()

    print("Calificaciones ordenadas de menor a mayor:", lista_notas)

    # Utilizar la función para sumar las calificaciones
    total_puntos = sumar_calificaciones(todas_las_notas)

    print("El total de puntos acumulados es:", total_puntos)


# ==========================================
# SECCIÓN 2: TRABAJO CON DICCIONARIOS
# ==========================================

# Función para buscar un contacto
def buscar_contacto(contactos, nombre):

    if nombre in contactos:
        return contactos[nombre]
    else:
        return None


def trabajar_diccionarios():

    # Crear un diccionario con tres contactos
    contactos = {
        "Ana": "555-0101",
        "Luis": "555-0102",
        "Mia": "555-0103"
    }

    print("\n--- AGENDA DE CONTACTOS ---")

    # Capturar un nuevo contacto
    nombre_nuevo = input("Ingresa el nombre del nuevo contacto: ")
    telefono_nuevo = input("Ingresa el teléfono del nuevo contacto: ")

    # Agregar el nuevo contacto
    contactos[nombre_nuevo] = telefono_nuevo

    # Mostrar los nombres de los contactos
    print("\nContactos registrados:")

    for nombre in contactos.keys():
        print(nombre)

    # Buscar un contacto
    nombre_buscar = input("\nIngresa el nombre que deseas buscar: ")

    telefono = buscar_contacto(contactos, nombre_buscar)

    if telefono is not None:
        print("El teléfono de", nombre_buscar, "es:", telefono)
    else:
        print("No se encontró ese contacto.")


# ==========================================
# SECCIÓN 3: MANEJO DE EXCEPCIONES
# ==========================================

def trabajar_excepciones():

    print("\n--- MANEJO DE EXCEPCIONES ---")

    try:

        # Pedir dos números enteros
        numero1 = int(input("Ingresa el primer número entero: "))
        numero2 = int(input("Ingresa el segundo número entero: "))

        # Realizar la suma
        suma = numero1 + numero2

        print("La suma de los dos números es:", suma)

        # Realizar la división
        try:

            resultado = numero1 / numero2

            print("La división del primer número entre el segundo es:", resultado)

        except ZeroDivisionError:

            print("Error: No es posible dividir entre cero.")
            print("Ingresa un divisor diferente de 0.")

    except ValueError:

        print("Error: Debes ingresar números enteros.")
        print("No puedes ingresar letras ni dejar el espacio vacío.")


# ==========================================
# SECCIÓN 4: TRABAJO CON STRINGS
# ==========================================

# Función para contar palabras
def contar_palabras(texto):

    palabras = texto.split()

    return len(palabras)


def trabajar_strings():

    print("\n--- ANALIZADOR DE STRINGS ---")

    # Crear una variable de texto
    mensaje = "Python es un lenguaje poderoso"

    # Mostrar el mensaje
    print("Mensaje original:", mensaje)

    # Mostrar la longitud del mensaje
    longitud = len(mensaje)

    print("Longitud del mensaje:", longitud)

    # Convertir el mensaje a mayúsculas
    mensaje_mayusculas = mensaje.upper()

    print("En mayúsculas:", mensaje_mayusculas)

    # Reemplazar una palabra
    mensaje_reemplazado = mensaje.replace("Python", "programación")

    print("Texto reemplazado:", mensaje_reemplazado)

    # Contar las palabras usando una función
    cantidad_palabras = contar_palabras(mensaje)

    print("Palabras totales:", cantidad_palabras)


# ==========================================
# MENÚ PRINCIPAL
# ==========================================

def menu_principal():

    while True:

        print("\n==========================================")
        print("             MENÚ PRINCIPAL")
        print("==========================================")
        print("1. Tuplas")
        print("2. Diccionarios")
        print("3. Excepciones")
        print("4. Strings")
        print("5. Finalizar")
        print("==========================================")

        opcion = input("Selecciona una opción: ")

        if opcion == "1":

            trabajar_tuplas()

        elif opcion == "2":

            trabajar_diccionarios()

        elif opcion == "3":

            trabajar_excepciones()

        elif opcion == "4":

            trabajar_strings()

        elif opcion == "5":

            print("\nPrograma finalizado.")
            print("Gracias por utilizar el programa.")

            break

        else:

            print("\nOpción no válida.")
            print("Selecciona una opción del 1 al 5.")


# ==========================================
# INICIO DEL PROGRAMA
# ==========================================

menu_principal()
```
