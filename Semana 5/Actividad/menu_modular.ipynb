En esta actividad realizamos un programa en Python utilizando tuplas,
diccionarios, excepciones y strings. También utilizamos funciones y un
menú principal para poder seleccionar cada sección del programa.

```python
# ==========================================
# ACTIVIDAD SEMANA 5 - MENU MODULAR
# Nombre: Evelyn Itzel León Uribe
# Matrícula: AL07311179
# ==========================================


# ==========================================
# SECCIÓN 1: TRABAJO CON TUPLAS
# ==========================================

# Función que recibe la tupla y retorna la suma de sus elementos
def sumar_elementos(tupla_numeros):
    return sum(tupla_numeros)


def trabajar_tuplas():
    # Crear la tupla requerida 'numeros' con 5 elementos numéricos
    numeros = (85, 90, 78, 92, 88)

    print("\n--- TRABAJO CON TUPLAS ---")
    print("Tupla inicial (numeros):", numeros)

    # Acceder e imprimir el tercer elemento (índice 2)
    print("El tercer elemento de la tupla es:", numeros[2])

    # Capturar dos números adicionales
    try:
        num1 = float(input("Ingresa el primer número adicional: "))
        num2 = float(input("Ingresa el segundo número adicional: "))
        
        # Anexarlos para crear una nueva tupla
        nueva_tupla = numeros + (num1, num2)
        print("Nueva tupla con los números agregados:", nueva_tupla)

        # Convertir la tupla a lista, ordenar y mostrar
        lista_numeros = list(nueva_tupla)
        lista_numeros.sort()
        print("Lista ordenada de menor a mayor:", lista_numeros)

        # Usar la función para sumar y retornar el total
        total_suma = sumar_elementos(nueva_tupla)
        print("La suma de todos los elementos es:", total_suma)

    except ValueError:
        print("Error: Debes ingresar valores numéricos válidos.")


# ==========================================
# SECCIÓN 2: TRABAJO CON DICCIONARIOS
# ==========================================

# Función para buscar y retornar el teléfono de un contacto
def buscar_contacto(contactos, nombre):
    if nombre in contactos:
        return contactos[nombre]
    else:
        return None


def trabajar_diccionarios():
    # Crear el diccionario 'contactos' con 3 registros iniciales
    contactos = {
        "Ana": "555-0101",
        "Luis": "555-0102",
        "Mia": "555-0103"
    }

    print("\n--- AGENDA DE CONTACTOS ---")

    # Capturar y agregar un nuevo contacto
    nombre_nuevo = input("Ingresa el nombre del nuevo contacto: ")
    telefono_nuevo = input("Ingresa el teléfono del nuevo contacto: ")
    contactos[nombre_nuevo] = telefono_nuevo

    # Iterar sobre las claves para imprimir solo los nombres
    print("\nNombres de los contactos registrados:")
    for nombre in contactos.keys():
        print("-", nombre)

    # Utilizar la función de búsqueda
    nombre_buscar = input("\nIngresa el nombre que deseas buscar: ")
    telefono = buscar_contacto(contactos, nombre_buscar)

    if telefono is not None:
        print(f"El teléfono de {nombre_buscar} es: {telefono}")
    else:
        print(f"No se encontró el contacto '{nombre_buscar}'.")


# ==========================================
# SECCIÓN 3: MANEJO DE EXCEPCIONES
# ==========================================

def trabajar_excepciones():
    print("\n--- MANEJO DE EXCEPCIONES ---")

    try:
        # Solicitar dos números enteros
        numero1 = int(input("Ingresa el primer número entero: "))
        numero2 = int(input("Ingresa el segundo número entero: "))

        # Calcular y mostrar la suma
        suma = numero1 + numero2
        print("La suma de los dos números es:", suma)

        # Bloque específico para la división y ZeroDivisionError
        try:
            resultado = numero1 / numero2
            print("La división del primer número entre el segundo es:", resultado)
        except ZeroDivisionError:
            print("Error: No es posible dividir entre cero. Ingresa un divisor diferente de 0.")

    except ValueError:
        print("Error: Debes ingresar números enteros válidos. No se permiten letras ni espacios vacíos.")


# ==========================================
# SECCIÓN 4: TRABAJO CON STRINGS
# ==========================================

# Función para contar palabras en un string
def contar_palabras(texto):
    palabras = texto.split()
    return len(palabras)


def trabajar_strings():
    print("\n--- ANALIZADOR DE STRINGS ---")

    # Crear la variable de texto 'mensaje'
    mensaje = "Python es un lenguaje poderoso"
    print("Mensaje original:", mensaje)

    # Imprimir la longitud
    print("Longitud del mensaje:", len(mensaje))

    # Transformar a mayúsculas
    mensaje_mayusculas = mensaje.upper()
    print("En mayúsculas:", mensaje_mayusculas)

    # Buscar y reemplazar una palabra clave
    mensaje_reemplazado = mensaje.replace("Python", "programación")
    print("Texto reemplazado:", mensaje_reemplazado)

    # Función que retorna la cantidad de palabras
    cantidad_palabras = contar_palabras(mensaje)
    print("Palabras totales en el mensaje:", cantidad_palabras)


# ==========================================
# MENÚ PRINCIPAL INTERACTIVO
# ==========================================

def menu_principal():
    while True:
        print("\n==========================================")
        print("             MENÚ PRINCIPAL")
        print("==========================================")
        print("1. Tuplas")
        print("2. Diccionarios")
        print("3. Excepciones")
        print("4. Strings")
        print("5. Finalizar")
        print("==========================================")

        opcion = input("Selecciona una opción (1-5): ")

        if opcion == "1":
            trabajar_tuplas()
        elif opcion == "2":
            trabajar_diccionarios()
        elif opcion == "3":
            trabajar_excepciones()
        elif opcion == "4":
            trabajar_strings()
        elif opcion == "5":
            print("\nPrograma finalizado.")
            print("Gracias por utilizar el programa.")
            break
        else:
            print("\nOpción no válida. Por favor selecciona una opción del 1 al 5.")


# ==========================================
# INICIO DEL PROGRAMA
# ==========================================

if __name__ == "__main__":
    menu_principal()
```
